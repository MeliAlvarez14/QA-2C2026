```python
%%file testSimpsons.py
import pytest
import requests


def testInfoPersonajeMaggieExitoso():
  response = requests.get('https://thesimpsonsapi.com/api/characters/5')
  genero = response.json()['gender']
  nombre = response.json()['name']
  edad = response.json()['age']

  assert response.status_code == 200
  assert 'Female' == genero
  assert 'Maggie Simpson' == nombre
  assert edad == 1

def testPersonajeNoEncontrado():
  response = requests.get('https://thesimpsonsapi.com/api/characters/9999')
  assert response.status_code == 404

def testPersonajesPorPaginacion():

    filtro = {"page": 1}
    response = requests.get('https://thesimpsonsapi.com/api/characters', params=filtro)
    data = response.json()

    assert response.status_code == 200

    if isinstance(data, list):
        assert len(data) <= 20
        assert "name" in data[0]
    elif isinstance(data, dict):
        items = data.get("results", data.get("data", []))
        assert len(items) <= 20
```
