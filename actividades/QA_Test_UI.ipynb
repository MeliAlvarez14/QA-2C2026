#Tests

```python
%%file testCafe.py
from playwright.sync_api import sync_playwright
import pytest

@pytest.fixture(scope="function")
def navegador():
    with sync_playwright() as p:
        browser = p.chromium.launch(headless=True)
        page = browser.new_page()
        page.goto("https://coffee-cart.app/")
        yield page
        browser.close()

def testAgregarProducto(navegador):
    navegador.locator("div.cup-body").first.click()

    checkout_button = navegador.locator("button.pay")

    assert "Total" in checkout_button.inner_text()
    assert "$10.00" in checkout_button.inner_text()

def testSumaMultiplesProductos(navegador):
    cafes = navegador.locator("div.cup-body")

    cafes.nth(0).click()
    cafes.nth(1).click()
    cafes.nth(3).click()

    checkout_button = navegador.locator("button.pay")

    assert "$30.00" in checkout_button.inner_text()

def testProcesoCompraExitosa(navegador):
  navegador.locator("div.cup-body").first.click()

  checkout_button = navegador.locator("button.pay")
  checkout_button.click()

  navegador.locator("#name").fill("Melina")
  navegador.locator("#email").fill("melina@gmail.com")

  navegador.locator("#submit-payment").click()

  success_message = navegador.locator(".snackbar, .success, div[role='alert']")
  assert "Total: $0.00" in navegador.locator("button.pay").inner_text()
```
