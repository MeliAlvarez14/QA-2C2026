```python
def divisible(numero, divisor):
    return numero % divisor == 0

def fizzbuzz(numero):
    if divisible(numero, 3) and divisible(numero, 5):
        return 'FizzBuzz'
    if divisible(numero, 3):
        return 'Fizz'
    if divisible(numero, 5):
        return 'Buzz'
    return numero

def listarN(N):
    lista = []
    for i in range(1, N+1):
        lista.append(fizzbuzz(i))
    return lista
```

```python
import unittest

class FizzBuzzTest(unittest.TestCase):

    def testFizzbuzz(self):
        self.assertEqual('FizzBuzz', fizzbuzz(30))

    def testFizz(self):
        self.assertEqual('Fizz', fizzbuzz(12))

    def testBuzz(self):
        self.assertEqual('Buzz', fizzbuzz(25))

    def testNumero(self):
        self.assertEqual(1, fizzbuzz(1))

unittest.main(defaultTest='FizzBuzzTest', argv=[''], verbosity=2, exit=False)
```
